### 1.1 Álgebra

#### Resolver una ecuación algebraica con una incógnita

```python
# Cargar librería con funciones específicas
from sympy import symbols, Eq, solve

# Definir la incógnita
x = symbols('x')

# Definir la ecuación
ecuacion = Eq(2*x + 3, 15)

# Resolver la ecuación
solucion = solve(ecuacion, x)
print("Solución para la ecuación 2x + 3 = 15: x =", solucion)
```

#### Resolver un sistema de ecuaciones de 2 incógnitas

```python
# Utilizamos la misma librería y funciones por lo que no hay que volver a cargar

# Definir las incógnitas
x, y = symbols('x y')

# Definir las ecuaciones
ecuacion1 = Eq(2*x + 3*y, 6)
ecuacion2 = Eq(-x + y, 2)

# Resolver el sistema de ecuaciones
soluciones = solve((ecuacion1, ecuacion2), (x, y))
print("Soluciones para el sistema de ecuaciones:")
print(f"x = {soluciones[x]}")
print(f"y = {soluciones[y]}")
```

#### Resolver un sistema de ecuaciones lineales de mayor tamaño

```python
# Cargar librería con funciones específicas
from sympy import Matrix, linsolve

# Definir el sistema de ecuaciones como matrices
A = Matrix([[2, 3, 1], [-1, 1, 0], [3, -2, -1]])
b = Matrix([11, 1, -4])

print(A[0,:],b[0])
print(A[1,:],b[1])
print(A[2,:],b[2])

# Resolver el sistema de ecuaciones
sistema = (A, b)
soluciones = linsolve(sistema)
print("Soluciones para el sistema de ecuaciones lineales:")
for solucion in soluciones:
    print(solucion)
```

#### Composición de Funciones

```python
def f(x):
    return 2*x + 3
    
def g(x):
    return x**2

# Composición de funciones: h(x) = f(g(x)) = 2(x^2) + 3
def h(x):
    return f(g(x))

# Valores de y para la composición
y_h = h(x)

# Graficar la composición de funciones
plt.figure(figsize=(6, 6))
plt.plot(x, y_h, label='h(x) = 2(x^2) + 3',linestyle = '--', linewidth = 2)
plt.title('Composición de Funciones')
plt.xlabel('x')
plt.ylabel('h(x)')
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(color = 'gray', linestyle = '--', linewidth = 1)
plt.legend()
plt.show()
```

#### Función inversa

```python
# Función original: f(x) = x + 2
def f(x):
    return x + 2

# Función inversa: f⁻¹(x) = x - 2
def f_inversa(x):
    return x - 2

# Valores de x
x = np.linspace(-10, 10, 200)

# Graficar la función y su inversa
plt.figure(figsize=(6, 6))
plt.plot(x, f(x), label='f(x) = x + 2', linewidth=2)
plt.plot(x, f_inversa(x), label='f⁻¹(x) = x - 2',
         linestyle='--', linewidth=2)
plt.plot(x, x, label='y = x', color='gray', linestyle=':')
plt.title('Función y su inversa')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.xlim(-10, 10)
plt.ylim(-10, 10)
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()
```

### 2.4 Condición de Ortogonalidad

```python
v1 = np.array([2, 1])
v2 = np.array([-1, 2])
producto_interno = np.dot(v1, v2)
print("Producto interno: ", producto_interno)
print("Ortogonalidad: ", "Ortogonales" if producto_interno == 0 else "No ortogonales")
```

```python
# Graficar los vectores desde el origen
fig, ax = plt.subplots(figsize=(6, 6))

ax.quiver(0, 0, v1[0], v1[1],
          angles='xy', scale_units='xy', scale=1,
          color='blue', label='v₁ = (2, 1)')

ax.quiver(0, 0, v2[0], v2[1],
          angles='xy', scale_units='xy', scale=1,
          color='red', label='v₂ = (-1, 2)')

# Marcar el ángulo recto entre los vectores
u1 = v1 / np.linalg.norm(v1)
u2 = v2 / np.linalg.norm(v2)

marca = 0.3 * np.array([u1, u1 + u2, u2])
ax.plot(marca[:, 0], marca[:, 1], color='black')

# Configurar la gráfica
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_aspect('equal', adjustable='box')

ax.set_title('Vectores ortogonales: v₁ · v₂ = 0')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(linestyle='--', alpha=0.5)
ax.legend()

plt.show()
```
