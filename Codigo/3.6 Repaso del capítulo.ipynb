**Datos:** Iris contiene medidas en centímetros de sépalos y pétalos de tres especies: setosa, versicolor y virginica.

- `sepal_length` Longitud del sépalo (cm)
- `sepal_width` Ancho del sépalo (cm)
- `petal_length` Longitud del pétalo (cm)
- `petal_width` Ancho del pétalo (cm)
- `species` Especie

```python
#Cargar bibliotecas esenciales para el análisis de datos
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib

print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
print("seaborn:", sns.__version__)
print("Matplotlib:", matplotlib.__version__)
```

```python
iris = sns.load_dataset("iris")
```

### 2. Calcula estadísticas descriptivas

#### Convierte las cuatro columnas numéricas originales en un arreglo de NumPy.

```python
sepal_l = np.array(iris.sepal_length)
sepal_w = np.array(iris.sepal_width)
petal_l = np.array(iris.petal_length)
petal_w = np.array(iris.petal_width)
```

#### Calcula la media, mediana, desviación estándar poblacional, mínimo y máximo de cada columna.

```python
def calculadora(columna):
    print("Media: ",np.mean(columna))
    print("Mediana: ",np.median(columna))
    print("Desviación estándar poblacional: ",np.std(columna))
    print("Mínimo: ",np.min(columna))
    print("Máximo: ",np.max(columna))
```

```python
calculadora(sepal_l)
```

```python
calculadora(sepal_w)
```

```python
calculadora(petal_l)
```

```python
calculadora(petal_w)
```

#### Obtén los mismos resultados pero con Pandas y con redondeo de dos decimales

```python
np.round(iris.describe(),2)
```

### 3. Filtra y ordena observaciones

#### Selecciona las flores de la especie virginica con longitud de pétalo mayor que 5.5cm

```python
iris[(iris.species == "virginica") & (iris.petal_length > 5.5)]
```

#### Muestra únicamente species, petal_length y petal_width

```python
iris[["species","petal_length","petal_width"]]
```

#### Ordena los resultados de mayor a menor longitud del pétalo

```python
iris.sort_values(by=['petal_length'], ascending=False)
```

### 4. Crea nuevas variables

#### Crea petal_ratio: longitud del pétalo dividida entre su ancho.

```python
iris['petal_ratio']=iris.petal_length/iris.petal_width 
iris.head()
```

#### Crea petal_size: Largo si petal_length es mayor o igual a 4cm; corto en los demás casos.

```python
iris['petal_size'] = np.where(iris['petal_length']>=4, "Largo", "Corto")

iris.head()
```

### 5. Compara las especies

#### Agrupa los datos por especie. Calcula media, mínimo y máximo de petal_length y petal_width.

```python
iris.groupby('species')[['petal_length','petal_width']].mean()
```

```python
iris.groupby('species')[['petal_length','petal_width']].min()
```

```python
iris.groupby('species')[['petal_length','petal_width']].max()
```

#### Ordena las especies según su longitud media del pétalo

```python
table = iris.groupby('species')[['petal_length']].mean()
table.sort_values(by="petal_length", ascending=True)
```

### 6. Construye histogramas

#### Crea una figura con dos subgráficas: un histograma de sepal_length y otro de petal_length.
#### Utiliza 15 intervalos en ambos histogramas
#### Incluye títulos, etiquetas con unidades y cuadrícula

```python
plt.hist(iris.sepal_length, bins= 15, color = 'green', alpha = 0.5,label='sepal length')
plt.hist(iris.petal_length, bins= 15, color = 'navy', alpha = 0.5,label='petal length')
plt.title('Longitud de Sépalo vs Pétalo')
plt.xlabel('Longitud (cm)')
plt.ylabel('Frecuencia')
plt.legend()
plt.grid()
plt.show()
```

### 7. Explora relaciones entre variables

#### Grafica petal_length en el eje horizontal y petal_width en el vertical
#### Representa cada especie con un color diferente.
#### Incluye leyenda, título y etiquetas con unidades.

```python
sns.scatterplot(data=iris, x='petal_length', y='petal_width', hue="species", style='species')
plt.title('Scatterplot')
plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
plt.grid()
plt.show()
```

### 9. Analiza las correlaciones

#### Selecciona úniamente las cuatro medidas originales y calcula la matriz de correlación de Pearson.
#### Representa la matriz mediante un mapa de calor.
#### Muestra coeficientes con dos decimales y una escala de color de -1 a 1

```python
iris = iris.drop(columns=['petal_ratio','petal_size'])
iris.head()
```

```python
matrix = iris.select_dtypes(include='number').corr()
sns.heatmap(matrix, annot=True, fmt='.2f', vmin = -1, vmax=1)
```

#### Elige dos variables que permitan distinguir visualmente las especies.
#### Grafica las variables y marca el promedio de cada especie con una estrella de mayor tamaño.

```python
iris.groupby('species')[['petal_length','petal_width']].mean()
```

```python
table = iris.groupby('species')[['petal_length','petal_width']].mean()
proms = np.array(table)

sns.scatterplot(data=iris, x='petal_length', y='petal_width', hue="species", style='species')
plt.scatter(proms[0,0], proms[0,1], marker='*', s=150, color='blue', edgecolor='black')
plt.scatter(proms[1,0], proms[1,1], marker='*', s=150, color='orange', edgecolor='black')
plt.scatter(proms[2,0], proms[2,1], marker='*', s=150, color='green', edgecolor='black')
plt.title('Scatterplot')
plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
plt.grid()
plt.show()
```
