##### 4.1 Límites

##### Importar las Bibliotecas Necesarias

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

##### Límites

```python
# Definir la función
def f(x):
    return (x**2 - 1) / (x - 1)

# Crear valores para x cerca del punto de interés (x=1)
x_values = np.linspace(0.9, 1.1, 1000)
y_values = f(x_values)

# Graficar la función
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='f(x) = (x^2 - 1) / (x - 1)')
plt.axvline(x=1, color='r', linestyle='--', label='x=1')
plt.axhline(y=2, color='g', linestyle='--', label='Límite en x=1')
plt.title('Límites')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Definir la variable simbólica
x = sp.symbols('x')

# Definir la función
f = sp.sin(x) / x

# Calcular el límite cuando x tiende a 0
limite = sp.limit(f, x, 0)

# Mostrar el resultado
print("El límite de sin(x)/x cuando x tiende a 0 es:", limite)
```

#### 4.2 Coeficiente Diferencial y Derivadas

##### Derivadas

```python
import sympy as sp

x = sp.symbols('x')
fx = x**2 + 2*x
derivada = sp.diff(fx, x)
print(derivada)
```

```python
fx = sp.sin(x)
derivada = sp.diff(fx, x)
print(derivada)
```

```python
fx = sp.log(x)
derivada = sp.diff(fx, x)
print(derivada)
```

##### Regla de la Potencia

```python
# Definir la variable y la función
f = x**5

# Derivada usando la regla de la potencia
f_prime = sp.diff(f, x)
print("Regla de la Potencia: ", f_prime)
```

##### Regla del Producto

```python
f = (30*x)*(x**4)
f_prime = sp.diff(f, x)
print("Regla del Producto", f_prime)
```

```python
u = 5*x**2 + 3*x - 2
v = sp.sin(6*x)

f = u*v
f_prime = sp.diff(f, x)

print("Función:", f)
print("Derivada:", f_prime)
```

##### Regla del Cociente

```python
numerador = 3*x**2 + 2*x + 1
denominador = x**2 + 1
h = numerador / denominador

# Derivada usando la regla del cociente
h_derivada = sp.diff(h, x)
print("La derivada de h(x) es:", h_derivada)
```

##### 4.3 Método Diferencial

```python
x_log = np.linspace(0.05, 5, 1000)
x_exp = np.linspace(-3, np.log(5), 1000)

plt.figure(figsize=(8, 8))
plt.plot(x_log, np.log(x_log), label='ln(x)')
plt.plot(x_exp, np.exp(x_exp), label='e^x')
# Referencia para observar la simetría de funciones inversas
plt.plot([-3, 5], [-3, 5], 'k--', label='y = x')

plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)
plt.xlim(-3, 5)
plt.ylim(-3, 5)
plt.gca().set_aspect('equal', adjustable='box')

plt.title('Logaritmo natural y exponencial')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```

##### 4.4 Diferencia entre Logaritmo y Exponencial

```python
# Definir la variable y las funciones
x = sp.symbols('x')
f_exp = 5*sp.exp(x)
f_log = 5*sp.log(x)

# Derivadas
f_exp_prime = sp.diff(f_exp, x)
f_log_prime = sp.diff(f_log, x)
print("Derivada de la Función Exponencial: ", f_exp_prime)
print("Derivada de la Función Logarítmica: ", f_log_prime)
```

##### 4.5 Derivada de la Composición de Funciones

```python
# Definir las variables y las funciones
x = sp.symbols('x')
g = x**2
f = sp.sin(g)

# Calcular la derivada de la composición usando la regla de la cadena
f_prime = sp.diff(f, x)

# Mostrar el resultado
print("La derivada de f(g(x)) = sin(x^2) es:", f_prime)
```

##### 4.6 Derivadas de Orden Superior y Derivadas Parciales

```python
# Definir la variable y la función
x = sp.symbols('x')
f = x**4

# Derivadas de orden superior
f_prime = sp.diff(f, x)   # Primera derivada
f_double_prime = sp.diff(f_prime, x)  # Segunda derivada
print("Primera Derivada: ", f_prime)
print("Segunda Derivada: ", f_double_prime)
```

```python
segunda_der = sp.diff(f, x, 2)
segunda_der
```

```python
# Definir las variables y la función
x, y = sp.symbols('x y')
f = x**2 + y**3

# Derivadas parciales
f_partial_x = sp.diff(f, x)
f_partial_y = sp.diff(f, y)
print("Derivada Parcial con respecto a x: ", f_partial_x)
print("Derivada Parcial con respecto a y: ", f_partial_y)
```

##### 4.7 Derivada de la Función Sigmoide

```python
# Definir la variable y la función sigmoide
x = sp.symbols('x')
sigmoid = 1 / (1 + sp.exp(-x))

# Derivada de la función sigmoide
sigmoid_prime = sp.diff(sigmoid, x)
print("Derivada de la Función Sigmoide: ", sigmoid_prime)
```

```python
# Función sigmoide y su derivada
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_prime(x):
    s = sigmoid(x)
    return s * (1 - s)

# Valores de x
x_values = np.linspace(-10, 10, 1000)
y_values = sigmoid(x_values)
y_prime_values = sigmoid_prime(x_values)

# Graficar la función sigmoide y su derivada
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='Sigmoid(x)')
plt.plot(x_values, y_prime_values, label="Sigmoid'(x)")
plt.title('Derivada de la Función Sigmoide')
plt.xlabel('x')
plt.ylabel('Valores')
plt.grid(True)
plt.legend()
plt.show()
```

#### Ejemplo cálculo de velocidad y aceleración de un automóvil

```python
# Generar datos de tiempo (en segundos)
t = np.linspace(1, 9, 1000)

# Simular datos de posición (en metros)
posicion = 3*t**2 + 2*t + 1

# Graficar la posición del automóvil
plt.figure(figsize=(10, 6))
plt.plot(t, posicion, label='Posición (m)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Posición (m)')
plt.title('Posición del Automóvil a lo Largo del Tiempo')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Calcular la velocidad (derivada de la posición)
velocidad = np.gradient(posicion, t)

# Graficar la velocidad del automóvil
plt.figure(figsize=(10, 6))
plt.plot(t, velocidad, label='Velocidad (m/s)', color='orange')
plt.xlabel('Tiempo (s)')
plt.ylabel('Velocidad (m/s)')
plt.title('Velocidad del Automóvil a lo Largo del Tiempo')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Calcular la aceleración (derivada de la velocidad)
aceleracion = np.gradient(velocidad, t)

# Graficar la aceleración del automóvil
plt.figure(figsize=(10, 6))
plt.plot(t, aceleracion, label='Aceleración (m/s²)', color='red')
plt.xlabel('Tiempo (s)')
plt.ylabel('Aceleración (m/s²)')
plt.title('Aceleración del Automóvil a lo Largo del Tiempo')
plt.grid(True)
plt.legend()
plt.show()
```
