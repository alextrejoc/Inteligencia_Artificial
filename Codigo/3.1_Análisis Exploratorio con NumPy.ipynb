### 4. Uso de argmax y argmin

#### Genera un arreglo aleatorio de 10 elementos

```python
np.random.seed(0)
arreglo = np.array(np.random.rand(10))

arreglo
```

#### Encuentra el índice del valor máximo y el índice del valor mínimo

```python
ind = np.argmax(arreglo)
arreglo = pd.Series(arreglo)
valor = arreglo.iloc[ind]
print("Indice mayor", ind)
print("Valor",valor)
```

```python
ind = np.argmin(arreglo)
arreglo = pd.Series(arreglo)
valor = arreglo.iloc[ind]
valor
print("Indice minimo", ind)
print("Valor",valor)
```

#### Cambia el valor máximo a 0 utilizando su índice

```python
arreglo[8] = 0
print(arreglo)
```

### 8. Estadísticas

#### Genera un arreglo de 100 elementos con números aleatorios entre 1 y 100

```python
arreglo = np.array(np.random.randint(1,101,100))
arreglo
```

#### Calcula la media, la mediana, el valor máximo, el valor mínimo y la desviación estándar

#### Encuentra el percentil 25, 50 y 75 de los datos

```python
arreglo = pd.Series(arreglo)
arreglo.describe()
```

```python
np.random.seed(8)
stat_array = np.random.randint(1,101,100)

#Calculo de medias, medianas, max...
media = np.mean(stat_array)
mediana = np.median(stat_array)
max = np.max(stat_array)
min = np.min(stat_array)
desviacion_est = np.std(stat_array)
desviacion_est_m = np.std(stat_array,ddof=1)
#Calculo de percentiles
percentil25 = np.percentile(stat_array,25)
percentil50 = np.percentile(stat_array,50)
percentil75 = np.percentile(stat_array,75)
print("Arreglo:", stat_array)
print("Media:", media)
print("Mediana:", mediana)
print("Máximo:", max)
print("Mínimo:", min)
print("Desviación estándar:", desviacion_est)
print("Desviación estándar muestra:", desviacion_est_m)
print("Percentil 25:", percentil25)
print("Percentil 50:", percentil50)
print("Percentil 75:", percentil75)
```

### 9. Concatenación y apilamiento

#### Crea dos arreglos de 1x5 y 5x1

```python
B = np.array(np.random.randint(0,10,5))
B = B.reshape((1,5))
print(B)
```

```python
C = np.array(np.random.randint(0,10,5))
C = C.reshape((5,1))
print(C)
```

#### Concátenalos en una sola fila y una sola columna

```python
np.hstack((B,C.T))
```

```python
np.vstack((B.T,C))
```

#### Aplica ambos arreglos con una matriz de 5x2

```python
np.concatenate((B.T, C), axis=1)
```
