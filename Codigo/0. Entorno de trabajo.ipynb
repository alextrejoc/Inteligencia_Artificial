#### Función Calculadora

```python
def calculadora(x,y):
    print("Suma: ", x + y)
    print("Resta: ", x - y)
    print("Multiplicación: ", x * y)
    print("División: ", x / y)
    print("Exponente: ", x ** y)
    print("Módulo: ", x % y)
```

```python
calculadora(10,4)
```

```python
calculadora(25,5)
```
